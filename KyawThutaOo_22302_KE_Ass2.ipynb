```python
# Remove "(" and ")" from coordinates
airbnb['coordinates'] = airbnb['coordinates'].str.replace("(","")
airbnb['coordinates'] = airbnb['coordinates'].str.replace(")","")
# Print the header of the column
airbnb['coordinates'].head()
```

```python
# Split column into two
lat_long = airbnb['coordinates'].str.split(",", expand = True)
lat_long.head()
```

```python
# Assign correct columns to latitude and longitude columns in airbnb
airbnb['latitude'] = lat_long[0]
airbnb['longitude'] = lat_long[1]
# Print the header and confirm new column creation
airbnb.head()
```

```python
# Convert latitude and longitude to float
airbnb['latitude'] = airbnb['latitude'].astype('float')
airbnb['longitude'] = airbnb['longitude'].astype('float')
# Print dtypes again
airbnb.dtypes
```

```python
# Drop coordinates column
airbnb.drop('coordinates', axis = 1, inplace = True)
```

```python
#reminder dataframe
airbnb.head()
```

```python
# Remove $ from price before conversion to float
airbnb['price'] = airbnb['price'].str.strip("$")
# Print dataframe to make sure change was done
display(airbnb['price'])
```

```python
# Convert price to float
airbnb['price'] = airbnb['price'].astype('float')
# Calculate mean of price after conversion
airbnb['price'].mean()
```

```python
#Task 1

# Calculate the maximum price
print(f"Maximum Price: {airbnb['price'].max()}")

# Calculate the minimum price
print(f"Minimum Price: {airbnb['price'].min()}")
```

```python
#Task 2
max_price = airbnb["price"].max()
min_price = airbnb["price"].min()

# Find the room type(s) with the maximum price and minimum price
max_room_types = airbnb.loc[airbnb["price"] == max_price, "room_type"]
min_room_types = airbnb.loc[airbnb["price"] == min_price, "room_type"]

print(f"Room type(s) with maximum price:\n{max_room_types}")
print(f"\nRoom type(s) with minimum price:\n{min_room_types}")
```

```python
#Task 3
# Categorize prices into five classes
labels = ["Budget", "Low", "Medium", "High", "Luxury"]

airbnb["price_category"] = pd.qcut(
    airbnb["price"], 5, labels=labels
)

print(airbnb["price_category"].value_counts())

print(
    airbnb.groupby("price_category")["price"]
    .agg(min_price="min", max_price="max")
)
```

```python
# Print header of column
airbnb['neighbourhood_full'].head()
# Split neighbourhood_full
borough_neighbourhood = airbnb['neighbourhood_full'].str.split(",", expand = True)
borough_neighbourhood.head()
# Create borough and neighbourhood columns
airbnb['borough'] = borough_neighbourhood[0]
airbnb['neighbourhood'] = borough_neighbourhood[1]
# Print header of columns
airbnb[['neighbourhood_full', 'borough', 'neighbourhood']].head()
# Drop neighbourhood_full column
airbnb.drop('neighbourhood_full', axis = 1, inplace = True)
# Print out unique values of borough and neighbourhood
print(airbnb['borough'].unique())
print(airbnb['neighbourhood'].unique())
# Strip white space from neighbourhood column
airbnb['neighbourhood'] = airbnb['neighbourhood'].str.strip()
# Print unique values again
print(airbnb['neighbourhood'].unique())
```

```python
#Task 4
#Find the noise from neighborhood column and remove it

# Display neighbourhood names before cleaning
print("Neighbourhoods before cleaning:")
print(airbnb["neighbourhood"].unique()[:10])

# Remove leading and trailing spaces
airbnb["neighbourhood"] = airbnb["neighbourhood"].str.strip()

# Display neighbourhood names after cleaning
print("\nNeighbourhoods after cleaning:")
print(airbnb["neighbourhood"].unique()[:10])

# Verify whether any leading or trailing spaces remain
remaining_noise = airbnb["neighbourhood"].str.startswith(" ") | airbnb["neighbourhood"].str.endswith(" ")

print("\nNumber of entries with extra spaces:", remaining_noise.sum())
```

```python
#Task 5
#Checking if there are any columns with rating>5
high_ratings = airbnb[airbnb["rating"] > 5]

# Display the number of invalid ratings
print("Listings with rating greater than 5:", len(high_ratings))

# Show the invalid listings
if not high_ratings.empty:
    print("\nListings with ratings above 5:")
    print(high_ratings[["listing_id", "rating"]])
else:
    print("\nNo listings have ratings greater than 5.")
```
